```python
import numpy as np
```

## Part 1 — Sorting

```python
# Sample Data
Weekly_sales = np.array([45, 12, np.nan, 30, 8, 45, 22])
Product_codes = np.array([105, 102, 105, 108, 102, 110, 105])
Product_names = np.array(["Kettle", "Mixer", "Toaster", "Kettle", "Blender"])
```

### Tasks
1. Arrange `weekly_sales` in ascending order without changing the original array. Print the result.
2. Make a copy of `weekly_sales` called `sales_copy`. Arrange `sales_copy` in ascending order so that the change is permanent on `sales_copy` itself. Print both `sales_copy` and the original `weekly_sales` afterward.
3. Produce a version of `weekly_sales` arranged from highest to lowest.
4. Get the index positions that would arrange `weekly_sales` in ascending order, and print them.
5. From `product_codes`, print only the distinct product codes (no repeats), along with how many distinct codes there are.
6. Print `product_names` arranged in alphabetical order.

```python
# 1
Asc = np.sort(Weekly_sales)
print(Asc)
```

```python
# 2
Sales_Copy = Weekly_sales.copy()
print("Copy of Weekely Sales (Sales Copy) -",Sales_Copy)

Asc = np.sort(Sales_Copy)
print("Asc of Sales Copy -",Asc)

print("Sales Copy -",Sales_Copy)
print("Weekly Sales -",Weekly_sales)
```

```python
# 3
Dsc_Weekly_Sales = np.sort(Weekly_sales)[::-1]
print("Desceding Order of Weekly Sales -",Dsc_Weekly_Sales)
```

```python
# 4
Index_Position = np.argsort(Weekly_sales)
print("Index Position in Ascending Order -",Index_Position)
```

```python
# 5
Distinct_Product_Codes = np.unique(Product_codes)
print("Distinct Product Codes -",Distinct_Product_Codes)
```

```python
# 6
Asc_Names = np.sort(Product_names)
print("Product Names in Alphabetical Orders -",Asc_Names)
```

## Part 2 — Adding Data

```python
# Sample Data
Warehouse_A = np.array([
    [120, 135, 140],
    [80, 95, 90],
    [60, 55, 70]
])
Warehouse_B = np.array([
    [200, 210, 205],
    [150, 160, 155],
    [90, 85, 100]
])
New_month = np.array([[150], [100], [75]])
New_product_row = np.array([[40, 42, 39]])
```

### Tasks
7. Add `22` to the end of `weekly_sales` without changing the original array. Print the new result, then print the original `weekly_sales` again.
8. Place `18` into `weekly_sales` so that it becomes the 3rd element (position index 2), shifting the rest forward. Print the result.
9. Attach `new_month` onto `warehouse_A` as an additional column. Print the result and its shape.
10. Attach `new_product_row` onto `warehouse_A` as an additional row. Print the result and its shape.
11. Combine `warehouse_A` and `warehouse_B` so that all products from both warehouses appear as separate rows in one array. Print the result and its shape.
12. Combine `warehouse_A` and `warehouse_B` so that each warehouse's matching months sit side by side as extra columns. Print the result and its shape.

```python
# 7
Adding_To_Weekely_Sales = np.append(Weekly_sales,[22])
print("Value Add at the End -",Adding_To_Weekely_Sales)

print("Original Weekely Sales -",Weekly_sales)
```

```python
# 8
Adding_To_Weekely_Sales = np.insert(Weekly_sales,2,18)
print("Value Add at the Position 3 -",Adding_To_Weekely_Sales)
```

```python
# 9
Adding_Column = np.hstack((Warehouse_A,New_month))
print("New Data of Warehouse A -",Adding_Column)

Shape = Adding_Column.shape
print("The Shape of new Warehouse A -",Shape)
```

```python
# 10
Adding_NewRow = np.vstack((Warehouse_A,New_product_row))
print("New Data of Warehouse A -",Adding_NewRow)

Shape = Adding_NewRow.shape
print("The Shape of new Warehouse A -",Shape)
```

```python
# 11
Combining_Warehouses = np.concatenate((Warehouse_A,Warehouse_B),axis=0)
print("The Result of Combining Warehouses as a Seperate Row -",Combining_Warehouses)

Shape = Combining_Warehouses.shape
print("The Shape After Combining Warehouse -",Shape)
```

```python
# 12
Combining_Warehouses = np.concatenate((Warehouse_A,Warehouse_B),axis=1)
print("The Result of Combining Warehouses as a side by side as extra columns -",Combining_Warehouses)

Shape = Combining_Warehouses.shape
print("The Shape After Combining Warehouse -",Shape)
```

## Part 3 — Filtering
13. From `warehouse_A`, pull out only the values greater than `90`.
14. From `warehouse_A`, pull out only the values that are greater than `60` and, at the same time, less than `140`.
15. From `warehouse_A`, pull out only the values that are either less than `60` or greater than `130`.
16. Find the row/column positions in `warehouse_A` where the value is below `70`.
17. Build a new version of `warehouse_A` where every value below `70` is replaced with `0`, and every other value stays as it is.
18. Given:
Restock_flags = np.array([0, 3, 0, 0, 7, 2, 0])
19. Given:
Flagged_codes = np.array([102, 999])
Check `product_codes` against `flagged_codes` and print which entries in `product_codes` are matches. Then print the actual matching values (not just True/False).

```python
# Sample Data
Restock_flags = np.array([0, 3, 0, 0, 7, 2, 0])
Flagged_codes = np.array([102, 999])
```

```python
# 13
Filtering_1 = Warehouse_A[Warehouse_A > 90]
print("Values which greater than 90 are -",Filtering_1)
```

```python
# 14
Filtering_2 = Warehouse_A[(Warehouse_A > 90) & (Warehouse_A < 140)]
print("Values which greater than 90 and less than 140 are -",Filtering_2)
```

```python
# 15
Filtering_3 = Warehouse_A[(Warehouse_A < 60) | (Warehouse_A > 130)]
print("Values which either less than 60 or greater than 130 are -",Filtering_3)
```

```python
# 16
Index = np.where(Warehouse_A < 70)
Rows,Columns = Index
print("Rows is -",Rows)
print("Column is -",Columns)
```

```python
# 17
New_Version = np.where(Warehouse_A < 70 , 0,Warehouse_A)
print("New Version of Warehouse A is -",New_Version)
```

```python
# 19
Matching = Product_codes[np.isin(Product_codes,Flagged_codes)]
print("Matching Products are -",Matching)
```
