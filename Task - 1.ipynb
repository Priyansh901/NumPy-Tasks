## Part 4 — Mathematical Functions

19. **`np.add()` / `np.subtract()`** — A new shipment adds `[5, 10, 2, 8, 1]` units to `stock`. Compute the new stock levels. Then, 3 units of each product were found damaged and removed — subtract `3` from every product's stock using a single line (broadcasting).

21. **`np.min()` / `np.max()`** — Find the cheapest and most expensive product prices.

23. **`np.sqrt()` / `np.power()`** — A discount formula NovaMart uses is `sqrt(price) * 2` for calculating loyalty points. Apply it to `prices`. Separately, use `np.power()` to calculate the square of the `stock` array (used in a variance calculation elsewhere).
24. **`np.abs()`** — A stock adjustment log shows changes as `[-3, 5, -2, 0, -1]` (negative = removed, positive = added). Use `abs()` to get the magnitude of change per product, ignoring direction.

```python
# 17
Multiply = np.multiply(Prices,Stock)
print("The total inventory value per product is -",Multiply)
```

```python
# 18
Sum = np.sum(Multiply)
print("The total inventory value across all 5 products -",Sum)
```

```python
# 19
Array = np.array([5,10,2,8,1])
print("The New Shipment -",Array)

Add = np.add(Stock,Array)
print("The New Stock Level is -",Add)

Subtract = np.subtract(Stock,3)
print("The New stock after removing damaged Product -",Subtract)
```

```python
# 20
Mean = np.mean(Prices)
print("The Avg Price is -",Mean)

Median = np.median(Prices)
print("The Middle Price is -",Median)

Standard_Deviation = np.std(Prices)
print("The Standard Deviation is -",Standard_Deviation)

Variance = np.var(Prices)
print("The Variance is -",Variance)

## The product prices have considerable variation. Some products are inexpensive, while others are much more expensive.
```

```python
# 21
Min = np.min(Prices)
print("The Min Price is -",Min)

Max = np.max(Prices)
print("The Max Price is -",Max)
```

```python
# 22
Round = np.round(Prices)
print("Prices rounded to the nearest ₹1 -",Round)

Floor = np.floor(Prices)
print("Prices rounded down -",Floor)

Ceil = np.ceil(Prices)
print("Prices rounded up -",Ceil)
```

```python
# 23
Square_Root = np.sqrt(Prices)*2
print("Loyalty Points is -",Square_Root)

Power = np.power(Stock,2)
print("The square of the stock array -",Power)
```

```python
# 24
Array_2 = np.array([-3, 5, -2, 0, -1])
print("A stock adjustment log shows changes as -",Array_2)

Absolute = np.abs(Array_2)
print("The magnitude of change per product, ignoring direction -",Absolute)
```

```python
# 25
Prod = np.prod(Stock)
print("The Prod of Stock -",Prod)
```
