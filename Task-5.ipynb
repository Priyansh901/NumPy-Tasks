```python
import numpy as np
```

```python
## Company Data

Hub_North_Sales = np.array([
    [18,20,22,19,21,780,23,20,19,22],
    [12,10,11,13,np.nan,12,14,13,12,11],
    [35,38,36,40,42,39,37,41,38,36],
    [8,7,9,8,6,7,np.nan,8,9,7],
    [25,27,24,26,28,25,23,24,26,27],
    [5,6,4,5,7,6,5,4,6,5]
])

Hub_South_Sales = np.array([
    [15,17,16,18,15,14,16,17,15,16],
    [9,8,np.nan,10,9,7,8,9,10,8],
    [30,32,31,29,33,30,28,31,32,30],
    [6,5,7,6,np.nan,5,6,7,6,5],
    [20,22,21,19,23,20,18,21,22,20],
    [np.nan,3,4,3,5,4,3,4,3,4]
])
# each array: rows = 6 products, columns = 10 days

Cost_Price = np.array([120.0, 250.0, 60.0, 400.0, 150.0, 90.0])
Sell_Price = np.array([180.0, 349.99, 89.99, 560.0, 219.0, 129.99])
# one value per product

Hub_Stock = np.array([
    [15, 8, 40, 5, 20, 6],
    [10, 4, 35, 3, 18, 9]
])
# rows = 2 hubs, columns = 6 products

Return_Log_North = np.array([0, -2, 0, -1, 0, 0, -3, 0, -1, 0])
# units returned by customers each day at North hub (product 0 only, negative = returned)

Promo_Codes = np.array(["WELCOME5","FESTIVE","WELCOME5","FESTIVE","CLEARANCE","WELCOME5","BIGDEAL","FESTIVE","CLEARANCE","WELCOME5"])
Flagged_Promo = np.array(["BIGDEAL","HACKED99"])

New_Hub_Week = np.array([
    [14,16,15,17,14,13,15,16,14,15],
    [7,6,8,7,6,5,7,8,7,6],
    [28,30,29,27,31,28,26,29,30,28],
    [5,4,6,5,4,3,5,6,5,4],
    [18,20,19,17,21,18,16,19,20,18],
    [3,4,2,3,5,4,3,2,4,3]
])
# a 3rd hub's 10-day sheet, same layout as the other two
```

## Task-1
Print the shape, number of dimensions, and total element count of `hub_north_sales` and `hub_south_sales`. Then print the shape of `cost_price`, `sell_price`, `hub_stock`, and `new_hub_week`.

```python
Shape_1 = Hub_North_Sales.shape
Ndim_1 = np.ndim(Hub_North_Sales)
Total_Elements_1 = np.size(Hub_North_Sales)
print("Shape of Hub North Sales is -",Shape_1)
print("Number of Dimensions of Hub North Sales -",Ndim_1)
print("Total Element count of Hub North Sales -",Total_Elements_1)

print("-"*50)

Shape_2 = Hub_South_Sales.shape
Ndim_2 = np.ndim(Hub_South_Sales)
Total_Elements_2 = np.size(Hub_South_Sales)
print("Shape of Hub South Sales is -",Shape_2)
print("Number of Dimensions of Hub South Sales -",Ndim_2)
print("Total Element count of Hub South Sales -",Total_Elements_2)

print("-"*50)
Shape_3 = Cost_Price.shape
Shape_4 = Sell_Price.shape
Shape_5 = Hub_Stock.shape
Shape_6 = New_Hub_Week.shape
print("Shape of Cost Price -",Shape_3)
print("Shape of Sell Price -",Shape_4)
print("Shape of Hub Stock -",Shape_5)
print("Shape of New Hub Week -",Shape_6)
```

## Task-2
Find the value in `hub_north_sales` that is far larger than everything around it — a clear data-entry mistake — and print its exact row and column position.

```python
Max = np.nanmax(Hub_North_Sales)
print("The Outlier is - ",Max)
Position = np.where(Hub_North_Sales == Max)
print("The Position of Outlier is -",Position)
```

## Task-3
Make a copy of `hub_north_sales`. In that copy, replace the value you found in Task 2 with the missing-value marker, so it's treated the same as the other gaps.

```python
Hub_North_Sales_Copy = Hub_North_Sales.copy()
Replace_Outlier = np.where(Hub_North_Sales == Max,np.nan,Hub_North_Sales_Copy)
print("The New Data after replace the Outlier with the missing-value marker in Copy of Hub North Sales :-")
print(Replace_Outlier)
```

## Task-4
Count how many values are missing in your corrected `hub_north_sales` copy from Task 3, and how many are missing in `hub_south_sales` — two separate counts.

```python
Missing_Count_Copy = np.sum(np.isnan(Hub_North_Sales_Copy))
print("Total values are missing in  corrected hub_north_sales copy is -",Missing_Count_Copy)
Missing_Count_North = np.sum(np.isnan(Replace_Outlier))
print("Total values are missing in hub_north_sales copy is -",Missing_Count_North)
```

## Task-5
Make a cleaned copy of each hub's array. For every product (row) in each hub, fill its missing days with that row's own 10-day average, ignoring the gaps while calculating it. No missing values should remain in either result.

```python
Replacing_With_Mean_North = np.where(np.isnan(Replace_Outlier),np.nanmean(Replace_Outlier,axis = 0),Replace_Outlier)
print("Filling its missing days with that row's own 10-day average of Hub North Sales :-")
print(Replacing_With_Mean_North)

print("-"*50)

Hub_South_Sales_copy = Hub_South_Sales.copy()

Replacing_With_Mean_South = np.where(np.isnan(Hub_South_Sales_copy),np.nanmean(Hub_South_Sales_copy,axis = 0),Hub_South_Sales_copy)
print("Filling its missing days with that row's own 10-day average of Hub South Sales :-")
print(Replacing_With_Mean_South)
```

## Task-6
Using the two cleaned arrays from Task 5, calculate each hub's total units sold for every product (add up all 10 days per row). Print both results.

```python
Total_Units_Sold_North_Per_Product = np.sum(Replacing_With_Mean_North,axis=1)
for i in range(6):
    print(f"Total units sold for {i+1} Product of North is {Total_Units_Sold_North_Per_Product[i]}")

print("-"*50)

Total_Units_Sold_South_Per_Product = np.sum(Replacing_With_Mean_South,axis=1)
for i in range(6):
    print(f"Total units sold for {i+1} Product of South is {Total_Units_Sold_South_Per_Product[i]}")
```

## Task-7
North hub had some customer returns during this window. Subtract the total number of units returned (from `return_log_north`) from product 0's total in your North hub result from Task 6, to get its real net total. Print the corrected value.

```python
Total_Units_Sold_North_Per_Product_copy = Total_Units_Sold_North_Per_Product.copy()
Total_Units_Sold_North_Per_Product_copy[0] = Total_Units_Sold_North_Per_Product[0] - np.abs(np.sum(Return_Log_North))
for i in range(6):
    print(f"The Corrected Value of  {i+1} Product of North is {Total_Units_Sold_North_Per_Product_copy[i]}")
```

## Task-8
Work out each product's profit margin by subtracting `cost_price` from `sell_price`. Print the result.

```python
Product_Profit_Margin = Sell_Price - Cost_Price
for i in range(len(Product_Profit_Margin)):
    print(f"Profit Margin for Product {i+1} is - {np.ceil(Product_Profit_Margin[i])}")
```

## Task-9
Using the totals from Task 6 (with North's product 0 corrected as in Task 7) and the margins from Task 8, calculate each hub's total profit per product. Print both results.

```python
Total_ProfiT_Per_North = Total_Units_Sold_North_Per_Product_copy * Product_Profit_Margin
for i in range(len(Product_Profit_Margin)):
    print(f"The Total Profit of Product {i+1} of North Hub is - {Total_ProfiT_Per_North[i]}")

print("-"*50)

Total_ProfiT_Per_South = Total_Units_Sold_South_Per_Product * Product_Profit_Margin
for i in range(len(Product_Profit_Margin)):
    print(f"The Total Profit of Product {i+1} of South Hub is - {Total_ProfiT_Per_South[i]}")
```

## Task-10
For each hub separately, find which product generated the most profit that window, and print that product's row number and its profit value.

```python
Max_Profit_North = np.max(Total_ProfiT_Per_North)
print("The Max Profit From North Hub is -",Max_Profit_North)
Max_Profit_North_Position = np.argmax(Total_ProfiT_Per_North)
print("The Position of Max Profit From North Hub is -",Max_Profit_North_Position)

print("-"*50)

Max_Profit_South = np.max(Total_ProfiT_Per_South)
print("The Max Profit From South Hub is -",Max_Profit_South)
Max_Profit_South_Position = np.argmax(Total_ProfiT_Per_South)
print("The Position of Max Profit From South Hub is -",Max_Profit_South_Position)
```
