```python
import numpy as np
```

```python
# Sample Data

Store0_sales = np.array([
    [45,50,48,52,60,55,58],
    [30,28,32,35,33,31,29],
    [70,75,72,80,78,76,74],
    [20,22,19,21,25,23,20],
    [15,18,16,17,19,20,18]
])

Store1_sales = np.array([
    [55,60,58,62,65,63,61],
    [40,38,42,45,43,41,39],
    [90,95,92,540,98,96,94],
    [25,27,24,26,28,25,23],
    [np.nan,20,18,19,21,22,20]
])

Store2_sales = np.array([
    [35,38,36,40,42,39,37],
    [np.nan,np.nan,28,30,29,27,26],
    [60,58,62,65,63,61,59],
    [15,17,14,16,18,15,14],
    [10,12,11,13,np.nan,14,12]
])
# each array: rows = 5 products, columns = 7 days

Prices = np.array([25.50, 40.00, 15.75, 60.20, 33.10])
Stock = np.array([
    [12, 8, 25, 5, 30],
    [9, 15, 6, 22, 4],
    [18, 3, 40, 7, 11]
])
# rows = 3 stores, columns = 5 products

Discount_codes = np.array(["SAVE10","WELCOME","SAVE10","FLASH","WELCOME","SAVE10","FLASH20"])
Flagged_codes = np.array(["FLASH20","SCAM50"])

New_store_week = np.array([
    [42,45,40,44,46,43,41],
    [33,31,35,32,34,30,29],
    [65,68,63,70,67,66,64],
    [18,20,17,19,21,18,16],
    [22,24,20,23,25,22,21]
])
# a 4th store's weekly sheet, same layout as the others
```

## Task-1
Print the shape, number of dimensions, and total element count of `store0_sales`, `store1_sales`, and `store2_sales`. Then print the shape of `prices`, `stock`, and `new_store_week`.

```python
Shape_1 = Store0_sales.shape
Shape_2 = Store1_sales.shape
Shape_3 = Store2_sales.shape

Shape_4 = Prices.shape
Shape_5 = Stock.shape
Shape_6 = New_store_week.shape

Dimension_1 = Store0_sales.ndim
Dimension_2 = Store1_sales.ndim
Dimension_3 = Store2_sales.ndim

Total_Element_1 = Store0_sales.size
Total_Element_2 = Store1_sales.size
Total_Element_3 = Store2_sales.size

print("Shape of Store-1 Data is -",Shape_1)
print("Shape of Store-2 Data is -",Shape_2)
print("Shape of Store-3  Data is -",Shape_3)

print("-"*50)

print("Shape of Prices Data is -",Shape_4)
print("Shape of New Store Week Data is -",Shape_5)

print("-"*50)

print("Dimension of Store-1 Data is -",Dimension_1)
print("Dimension of Store-2 Data is -",Dimension_2)
print("Dimension of Store-3 Data is -",Dimension_3)

print("-"*50)

print("Total Element in Store 1 Data -",Total_Element_1)
print("Total Element in Store 2 Data -",Total_Element_2)
print("Total Element in Store 3 Data -",Total_Element_3)
```

## Task-2
Find the value in `store1_sales` that is far larger than everything around it — a clear data-entry mistake — and print its exact row and column position.

```python
Max_Value = np.nanmax(Store1_sales)

Rows, Column = np.where(Store1_sales == Max_Value)

print("Maximum Value from Store-2 Sales Data is -", Max_Value)
print("Row Index is -", Rows)
print("Column Index is -", Column)
```

## Task-3
Make a copy of `store1_sales`. In that copy, replace the value you found in Task 2 with the missing-value marker instead of a number, so it's treated the same as the other gaps in the data.

```python
Store_2_Sales = Store1_sales.copy()

Missing_Value = np.where(np.isnan(Store_2_Sales),Max_Value,Store_2_Sales)
print("The Store 1 Sales Data after inserting missing Values -",Missing_Value)
```

## Task-4
Count how many values are missing in `store0_sales`, in your corrected `store1_sales` copy from Task 3, and in `store2_sales` — three separate counts.

```python
Total_Nan_Values_1 = np.sum(np.isnan(Store0_sales))
print("Total Nan Values in Store 1 Sales Data is -",Total_Nan_Values_1)

Total_Nan_Values_2 = np.sum(np.isnan(Store_2_Sales))
print("Total Nan Values in Store 2 Sales Copy Data is -",Total_Nan_Values_2)

Total_Nan_Values_3 = np.sum(np.isnan(Store2_sales))
print("Total Nan Values in Store 3 Sales Data is -",Total_Nan_Values_3)
```

## Task-5
Make a cleaned copy of each of the three store arrays. For every product (row) in each store, fill its missing days with that row's own weekly average, ignoring the gaps while calculating it. No missing values should remain anywhere in any of the three results.

```python
Store_1_Sales = Store0_sales.copy()
Store_3_Sales = Store2_sales.copy()

Replacing_1 = np.where(np.isnan(Store_1_Sales),np.nanmean(Store_1_Sales),Store_1_Sales)
print("Store 1 Sales Data after replacing nan values with that row average -",Replacing_1)

Replacing_2 = np.where(np.isnan(Store_2_Sales),np.nanmean(Store_2_Sales),Store_2_Sales)
print("Store 2 Sales Data after replacing nan values with that row average -",Replacing_2)

Replacing_3 = np.where(np.isnan(Store_3_Sales),np.nanmean(Store_3_Sales),Store_3_Sales)
print("Store 1 Sales Data after replacing nan values with that row average -",Replacing_3)
```

## Task-6
Using the three cleaned arrays from Task 5, calculate each store's total weekly units sold for every product (add up all 7 days per row). Print all three results.

```python
Weekly_Unit_Sold_1 = np.sum(Replacing_1,axis=1)
for i in range(5):
    print(f"Total Weekly Units Sold of {i+1} Product Of Store-1 is - {Weekly_Unit_Sold_1[i]}")

print("-"*50)

Weekly_Unit_Sold_2 = np.sum(Replacing_2,axis=1)
for i in range(5):
    print(f"Total Weekly Units Sold of {i+1} Product Of Store-2 is - {Weekly_Unit_Sold_2[i]}")

print("-"*50)

Weekly_Unit_Sold_3 = np.sum(Replacing_3,axis=1)
for i in range(5):
    print(f"Total Weekly Units Sold for {i+1} Product Of Store-2 is - {Weekly_Unit_Sold_3[i]}")
```

## Task-7
Using the totals from Task 6, calculate each store's weekly revenue per product by pairing each product's total with its matching price from `prices`. Print all three revenue results.

```python
Revenue_Store_1 = Weekly_Unit_Sold_1 * Prices
Revenue_Store_2 = Weekly_Unit_Sold_2 * Prices
Revenue_Store_3 = Weekly_Unit_Sold_3 * Prices
for i in range(5):
    print(f"Total Revenue of {i+1} Product Of Store-1 is - {Revenue_Store_1[i]}")

print("-"*50)

for i in range(5):
    print(f"Total Revenue of {i+1} Product Of Store-1 is - {Revenue_Store_2[i]}")

print("-"*50)

for i in range(5):
    print(f"Total Revenue of {i+1} Product Of Store-1 is - {Revenue_Store_3[i]}")
```
